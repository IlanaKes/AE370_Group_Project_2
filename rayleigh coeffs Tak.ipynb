```python
#Cd based dampening coeff function

def air_damp_coeff(w, l, v0):
    Cd = 2.0
    A = w*l
    rho = 1.225 #kg/m^3
    
    return (rho*Cd*A*v0)/2


np.eye(4)
```

```python
rho_s = 7850 #kg/m^3
w_o = 10 #m
th = 0.5 #m
w_i = w_o - 2*th
I = (w_o**2 - w_i**2)/12
Area = w_o**2 - w_i**2
E = 200*10**9 #Pa
L = 200 #m
dratio1 = 0.02
dratio2 = dratio1*2
```

```python
def func(n):
    dx = L/n
    m_mix = rho_s*Area*np.eye(n-3)
    R = (E*I)/(dx**4)
    A = 6*np.eye(n-3)
    A[0, 1] = A[-1, -2] = -4
    A[0, 2] = A[-1, -3] = 1
    A[1, 0] = A[1, 2] = A[-2, -1] = A[-2, -3] = -4
    A[1, 3] = A[-2, -4] = 1
    for i in range(2, n-5):
        A[i, i-1] = A[i, i+1] = -4
        A[i, i-2] = A[i, i+2] = 1
    
    A = R*A
    wlis = []
    for i in range(n-3):
        wlis.append(np.sqrt(A[i, i]/m_mix[i,i]))
    

    from scipy.linalg import eig

    # Define the matrices A and B
    
# Calculate the generalized eigenvalues and eigenvectors
# w will contain the eigenvalues
# vr will contain the right eigenvectors, where each column vr[:,i] 
# corresponds to the eigenvalue w[i]
    w, vr = eig(m_mix, A)
    solnmix = np.array([dratio1, dratio2])
    mix = np.array([[1/w[0], w[0]], [1/w[1], w[1]]])/2
    a, b = np.linalg.solve(mix, solnmix)
    c = a*A + b*m_mix
# Print the results
    return c

# Verification (A @ vr[:, i] = w[i] * B @ vr[:, i])
    #i = 0
    #lhs = A @ vr[:, i]
    #rhs = w[i] * B @ vr[:, i]
    #print(f"\nVerification for eigenvalue {w[i]}:")
    #print(f"LHS (A*v): {lhs}")
    #print(f"RHS (λ*B*v): {rhs}")

func(10)
```
